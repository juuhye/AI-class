```python
import matplotlib.pyplot as plt

from sklearn import datasets, metrics
from sklearn.model_selection import train_test_split

digits = datasets.load_digits()
plt.imshow(digits.images[0], cmap=plt.cm.gray_r, interpolation='nearest')

n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

# digits
```

```python
from sklearn.neighbors import KNeighborsClassifier

# knn - n_neighbors = 6
knn = KNeighborsClassifier(n_neighbors=6)
# knn - n_neighbors = 5
knn_5 = KNeighborsClassifier(n_neighbors=5)

X_train, X_test, y_train, y_test = train_test_split(
    digits.data, digits.target, test_size=0.2
)
knn.fit(X_train, y_train)
knn_5.fit(X_train, y_train)

y_pred = knn.predict(X_test)
y_pred_5 = knn_5.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
scores_5 = metrics.accuracy_score(y_test, y_pred_5)

print("knn - n_neighbors = 6", scores)
print("knn - n_neighbors = 5", scores_5)
```

```python
from sklearn.metrics import classification_report

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import RandomForestClassifier

from sklearn.linear_model import LogisticRegression
```

```python
model_dtc = DecisionTreeClassifier()
model_dtc.fit(X_train, y_train)

pred_dtc = model_dtc.predict(X_test)

scores = metrics.accuracy_score(y_test, pred_dtc)

print(classification_report(y_test, pred_dtc))
print(scores)
```

```python
model_rc = RandomForestClassifier()
model_rc.fit(X_train, y_train)

pred_rc = model_rc.predict(X_test)

scores = metrics.accuracy_score(y_test, pred_rc)

print(classification_report(y_test, pred_rc))
print(scores)
```

```python
model_lr = LogisticRegression()
model_lr.fit(X_train, y_train)

pred_lr = model_lr.predict(X_test)

scores = metrics.accuracy_score(y_test, pred_lr)

print(classification_report(y_test, pred_lr))
print(scores)
```
