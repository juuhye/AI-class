```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/car.data")
print(data) # 컬럼 명 없음

```

```python
df = pd.DataFrame(data)

df.columns = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class']

df


```

```python
label_encoders = {}

columns = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class']

for column in columns:

    label_encoders[column] = LabelEncoder()

    df[column] = label_encoders[column].fit_transform(df[column])
```

```python
X = df.drop("class", axis=1)
y = df['class']


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import classification_report

from sklearn.ensemble import RandomForestClassifier

from sklearn.linear_model import LogisticRegression

from sklearn.neighbors import KNeighborsClassifier
```

```python
model_dtc = DecisionTreeClassifier()
model_dtc.fit(X_train, y_train)

pred_dtc = model_dtc.predict(X_test)

print(classification_report(y_test, pred_dtc))
```

```python
model_rc = RandomForestClassifier()
model_rc.fit(X_train, y_train)

pred_rc = model_rc.predict(X_test)

print(classification_report(y_test, pred_rc))
```

```python
model_lr = LogisticRegression()
model_lr.fit(X_train, y_train)

pred_lr = model_lr.predict(X_test)

print(classification_report(y_test, pred_lr))
```

```python
model_knn =KNeighborsClassifier()
model_knn.fit(X_train, y_train)

pred_knn = model_knn.predict(X_test)

print(classification_report(y_test, pred_knn))
```
