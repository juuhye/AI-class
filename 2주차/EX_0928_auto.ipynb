```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/auto-mpg.data", sep=r'\s+', header=None)
print(data) # 컬럼 명 없음

data.head()

```

```python
df = pd.DataFrame(data)

df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year', 'origin', 'car_name']

df
```

```python
label_encoders = {}

columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year', 'origin', 'car_name']

for column in columns:

    label_encoders[column] = LabelEncoder()

    df[column] = label_encoders[column].fit_transform(df[column])
```

```python
df.drop("car_name", axis=1)

X = df.drop("mpg", axis=1)
y = df['mpg']


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import classification_report

from sklearn.ensemble import RandomForestClassifier

from sklearn.linear_model import LogisticRegression

from sklearn.neighbors import KNeighborsClassifier
```

```python
model_dtc = DecisionTreeClassifier()
model_dtc.fit(X_train, y_train)

pred_dtc = model_dtc.predict(X_test)

print(classification_report(y_test, pred_dtc))
```

```python
model_rc = RandomForestClassifier()
model_rc.fit(X_train, y_train)

pred_rc = model_rc.predict(X_test)

print(classification_report(y_test, pred_rc))
```

```python
model_lr = LogisticRegression()
model_lr.fit(X_train, y_train)

pred_lr = model_lr.predict(X_test)

print(classification_report(y_test, pred_lr))
```

```python
model_knn =KNeighborsClassifier()
model_knn.fit(X_train, y_train)

pred_knn = model_knn.predict(X_test)

print(classification_report(y_test, pred_knn))
```
